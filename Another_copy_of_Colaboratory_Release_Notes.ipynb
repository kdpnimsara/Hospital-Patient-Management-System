````python
# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

# We import pandas to handle data tables.
import pandas as pd

# We import numpy for doing math operations.
import numpy as np

# FIXED: We import seaborn correctly as sns
import seaborn as sns

# We import matplotlib to customize our plots.
import matplotlib.pyplot as plt

# We import train_test_split to split data.
from sklearn.model_selection import train_test_split

# We import StandardScaler to balance feature scales.
from sklearn.preprocessing import StandardScaler

# We import KNeighborsClassifier for KNN algorithm.
from sklearn.neighbors import KNeighborsClassifier

# We import functions to measure model performance.
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report


# ============================================================
# 2. LOAD DATASET DIRECTLY FROM UCI REPOSITORY
# ============================================================

# We save the online data link here.
data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"

# We define the names for columns here.
attribute_names = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg',
                  'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']

# We download and read the dataset now.
ph = pd.read_csv(data_url, names=attribute_names)


# ============================================================
# 3. DATA PREPROCESSING & CLEANING
# ============================================================

# We replace question marks with empty values.
ph = ph.replace('?', np.nan)

# We drop rows with missing values now.
ph = ph.dropna()

# We convert all data to float type.
ph = ph.astype(np.float64)

# We change all disease levels to 1.
ph['target'] = ph['target'].replace([1, 2, 3, 4], 1)


# ============================================================
# 4. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# We print the top 5 data rows.
print("--- First 5 Records ---")
print(ph.head())

# We print rows and columns count here.
print("\nDataframe Dimensions:")
print(ph.shape)

# We check for any remaining missing values.
print("\n--- Missing Values Check ---")
print(ph.isnull().sum())

# We set the class count plot size.
plt.figure(figsize=(7, 4))

# We create a bar chart for target.
sns.countplot(x='target', data=ph, palette='magma')

# We set the main title here.
plt.title("Analysis of Target Class Counts")

# We label the horizontal bottom X-axis.
plt.xlabel("Status (0 = Healthy, 1 = Disease)")

# We label the vertical side Y-axis.
plt.ylabel("Number of Patients")

# We show the class chart on screen.
plt.show()


# ============================================================
# 5. DATASET SPLITTING
# ============================================================

# We remove target column to get features.
X_features = ph.drop('target', axis=1)

# We save the target column alone here.
y_target = ph['target']

# We split data into training and testing.
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_features, y_target, test_size=0.20, random_state=42
)


# ============================================================
# 6. FEATURE SCALING (ESSENTIAL FOR KNN)
# ============================================================

# We initialize the standard scaling tool.
scaler_tool = StandardScaler()

# FIXED: Used the correct initialized 'scaler_tool' object name here
X_train_scaled = scaler_tool.fit_transform(X_train_raw)

# FIXED: Used the correct initialized 'scaler_tool' object name here
X_test_scaled = scaler_tool.transform(X_test_raw)


# ============================================================
# 7. TRAINING THE KNN CLASSIFIER
# ============================================================

# We build a new KNN model with 7 neighbors.
knn_heart_model = KNeighborsClassifier(n_neighbors=7)

# We teach this model using our training data.
knn_heart_model.fit(X_train_scaled, y_train)


# ============================================================
# 8. MODEL PREDICTION
# ============================================================

# We predict the answers for our test data.
test_predictions = knn_heart_model.predict(X_test_scaled)


# ============================================================
# 9. PERFORMANCE EVALUATION
# ============================================================

# We find the total accuracy of the predictions.
model_accuracy = accuracy_score(y_test, test_predictions)

# We print the accuracy percentage on screen.
print(f"Classifier Accuracy Score: {model_accuracy * 100:.2f}%\n")

# We print a short header for the report.
print("--- Comprehensive Breakdown Report ---")

# We print the full detailed classification report.
print(classification_report(y_test, test_predictions))

# We set the graph size to 7x5.
plt.figure(figsize=(7, 5))

# We calculate the confusion matrix counts now.
summary_matrix = confusion_matrix(y_test, test_predictions)

# We draw a blue heatmap for visualization.
sns.heatmap(summary_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Disease', 'Has Disease'], yticklabels=['No Disease', 'Has Disease'])

# We label the horizontal bottom X-axis here.
plt.xlabel("Model Predictions")

# We label the vertical side Y-axis here.
plt.ylabel("Ground Truth")

# We give a title to this heatmap.
plt.title("Evaluation Matrix Heatmap")

# We show the finished heatmap on screen.
plt.show()


# ============================================================
# 10. ERROR RATE ANALYSIS FOR OPTIMAL K VALUE
# ============================================================

# We create an empty list for errors.
misclassification_rates = []

# We test K values from 1 to 20.
for current_k in range(1, 21):

    # We make a temporary model for testing.
    predictive_node = KNeighborsClassifier(n_neighbors=current_k)

    # We train this temporary model now.
    predictive_node.fit(X_train_scaled, y_train)

    # We guess labels for the test data.
    node_outputs = predictive_node.predict(X_test_scaled)

    # We add the error rate to list.
    misclassification_rates.append(np.mean(node_outputs != y_test))

# We set the line graph size here.
plt.figure(figsize=(7, 4))

# We plot the errors using green line.
plt.plot(range(1, 21), misclassification_rates, color='green', linestyle='solid',
         marker='s', markerfacecolor='red', markersize=7)

# We give a title to this chart.
plt.title("Error Metric Variation across K Values")

# We label the horizontal K axis.
plt.xlabel("Selected K Hyperparameter")

# We label the vertical error axis.
plt.ylabel("Calculated Mean Error Rate")

# We turn on the background grid lines.
plt.grid(True)

# We show the final chart on screen.
plt.show()
````
